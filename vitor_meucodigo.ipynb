```python
%%writefile physics_tool.py

"""
physics.tools.py

Módulo simples de fisica para cálculos básicos:
- velocidade média
- distância
- aceleração média

Este arquivo pode ser publicado no Github e importado em outros projetos.
"""

def calcular_velocidade(distancia: float, tempo: float) -> float:
  """
  Calcula a velocidade média.

  Fórmula:
  v = distancia / tempo

  Parâmetros:
    distancia (float): Distância percorrida (ex.: metros, km).
    tempo (float): Tempo gasto (ex.: segundos, horas). Deve ser diferente de 0.

  Retorno:
    float: Velocidade média (na unidade compatível com as entradas).
  """
  assert isinstance(distancia,(int, float)), "distancia deve ser um número (int ou float)"
  assert isinstance(tempo,(int, float)), "tempo deve ser um número (int ou float)"
  assert tempo != 0, "tempo não pode ser 0 (divisão por zero)"

  return float(distancia) / float(tempo)

def calcular_distancia(velocidade: float, tempo: float) -> float:
  """
  Calcula a distância percorrida.

  Fórmula:
  d = velocidade * tempo

  Parâmetros:
    velocidade (float): Velocidade média (ex.: m/s, km/h).
    tempo (float): Tempo (ex.: segundos, horas).

  Retorno:
    float: Distância percorrida (na unidade compatível com as entradas).
  """
  assert isinstance(velocidade,(int, float)), "velocidade deve ser um número (int ou float)"
  assert isinstance(tempo,(int, float)), "tempo deve ser um número (int ou float)"

  return float(velocidade) * float(tempo)

def calcular_aceleracao(velocidade_inicial: float, velocidade_final: float, tempo: float) -> float:
  """
  Calcula a aceleração média.

  Fórmula:
  a = (velocidade_final - velocidade_inicial) / tempo

  Parâmetros:
    velocidade_inicial (float): Velocidade inicial (vi).
    velocidade_final (float): Velocidade final (vf).
    tempo (float): Tempo do intervalo. Deve ser diferente de 0.

  Retorno:
    float: Aceleração média (na unidade compatível com as entradas).
  """
  assert isinstance(velocidade_inicial, (int, float)), "velocidade_inicial deve ser numérica"
  assert isinstance(velocidade_final, (int, float)), "velocidade_final deve ser numérica"
  assert isinstance(tempo, (int, float)), "tempo deve ser numérico"
  assert tempo != 0, "tempo não pode ser 0 (divisão por zero)"

  return (float(velocidade_final) - float(velocidade_inicial)) / float(tempo)
```
